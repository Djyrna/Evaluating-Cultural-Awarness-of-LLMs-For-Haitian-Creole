Entity Filling Code

```python
import csv

input_file = "Cultural_Prompts_Final_Creole.csv"
output_file = "Cultural_creole_prompts_results_large.csv"

with open(input_file, mode="r", newline="", encoding="utf-8") as infile, \
    open(output_file, mode="w", newline="", encoding="utf-8") as outfile:

    reader = csv.reader(infile)
    writer = csv.writer(outfile)

    for row in reader:
        for element in row:
            message = element
            chat_response = client.chat.complete(
            model= model,
            messages = [
                  {
                      "role": "user",
                      "content": message,
                  }
              ]
            )
            writer.writerow([chat_response.choices[0].message.content])
```

Cultural Haitian Entity filling

Loading csv

```python
import pandas as pd

df_haitian_cultural = pd.read_csv("Cultural_creole_prompts_results.csv", header=None)
df_food = pd.read_csv("foods.csv", header=None)
df_book = pd.read_csv("authors.csv", header=None)
df_loc = pd.read_csv("locations.csv", header=None)
df_music = pd.read_csv("Musical_groups.csv", header=None)
df_music_type = pd.read_csv("Music.csv", header=None)
df_heroes = pd.read_csv("Political_heros_figures.csv", header=None)
df_festival = pd.read_csv("Festivals.csv", header=None)

food = df_haitian_cultural.iloc[0:29]
food[0].str.lower()
book = df_haitian_cultural.iloc[29:47]
location = df_haitian_cultural.iloc[47:73]
music_group= df_haitian_cultural.iloc[73:95]
music_type = df_haitian_cultural.iloc[95:103]
heroes = df_haitian_cultural.iloc[103:124]
festival = df_haitian_cultural.iloc[124:]
```

Haitian entity generated

```python
def commun_df(generated,list):
  commun = generated[generated[0].isin(list[0])]
  return commun
```

```python
commun_food = commun_df(food,df_food)
commun_book = commun_df(book,df_book)
commun_location = commun_df(location,df_loc)
commun_music_group = commun_df(music_group,df_music)
commun_music_type = commun_df(music_type,df_music_type)
commun_heroes = commun_df(heroes,df_heroes)
commun_festival = commun_df(festival,df_festival)
commun_total =  pd.concat([commun_food, commun_book, commun_location, commun_music_group, commun_music_type, commun_heroes, commun_festival], ignore_index=True)
```

Haitian Specificity

```python
def specificity(commun,total):
  return len(commun)/len(total)
```

```python
haitian_specificity = {}
haitian_specificity["food"] = specificity(commun_food,food)
haitian_specificity["book"] = specificity(commun_book,book)
haitian_specificity["location"] = specificity(commun_location,location)
haitian_specificity["music_group"] = specificity(commun_music_group,music_group)
haitian_specificity["music_type"] = specificity(commun_music_type,df_music_type)
haitian_specificity["heroes"] = specificity(commun_heroes,heroes)
haitian_specificity["festival"] = specificity(commun_festival,festival)
haitian_specificity["total"] = specificity(commun_total,df_haitian_cultural)
```

```python
haitian_specificity
```

Haitian Diversity

```python
def diversity(commun,total):
  return len(commun[0].unique())/len(total)
```

```python
haitian_diversity = {}
haitian_diversity["food"] = diversity(commun_food,df_food)
haitian_diversity["book"] = diversity(commun_book,df_book)
haitian_diversity["location"] = diversity(commun_location,df_loc)
haitian_diversity["music_group"] = diversity(commun_music_group,df_music)
haitian_diversity["music_type"] = diversity(commun_music_type,df_music_type)
haitian_diversity["heroes"] = diversity(commun_heroes,df_heroes)
haitian_diversity["festival"] = diversity(commun_festival,df_festival)
haitian_diversity["total"] = diversity(commun_total,df_haitian_cultural)
```
