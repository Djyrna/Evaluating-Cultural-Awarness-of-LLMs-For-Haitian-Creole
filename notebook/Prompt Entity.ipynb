Prompts for haitian creole Entity Filling

```python
from openpyxl import load_workbook, Workbook
import csv
from collections import defaultdict
input_file = "Prompts_Cultural_creole.xlsx"
output_file = "Prompts_cultural_final_creole.csv"

category_map = {
    "Food": "manje",
    "Authors": "otè",
    "Location": "lye",
    "Music": "Mizik",
    "Musical groups": "gwoup mikiz / chantè",
    "Festivals": "fèt",
    "Heroes/Political figures" : "ewo/ figi politik"
}

# Load Excel
wb = load_workbook(input_file)
ws = wb.active

# Write CSV
with open(output_file, mode="w", newline="", encoding="utf-8") as outfile:
    writer = csv.writer(outfile)

    # Loop through each row (skip header)
    for row in ws.iter_rows(min_row=2, min_col=1, max_col=2, values_only=True):
        category, element = row
        #print(category, element)

        if category and element:
            # Translate category
            category_creole = category_map.get(category, category)


        sentence = (
            f"Tanpri, nan fraz sa a ranplase [MASK] pa yon  entite ki soti nan kategori {category_creole} epi ban mwen egzakteman yon sèl antite kòm repons. Pa rekopye fraz la, pa ajoute okenn eksplikasyon oswa detay nan repons lan: "
            f"{element}"
        )

        writer.writerow([sentence])
```
