Prompts for story generation



```python
import csv

input_file = "/content/haitian_people_names.csv"
output_file = "Haitian_Name_story_prompts.csv"

with open(input_file, mode="r", newline="", encoding="utf-8") as infile, \
     open(output_file, mode="w", newline="", encoding="utf-8") as outfile:

    reader = csv.reader(infile)
    writer = csv.writer(outfile)

    for row in reader:
        for element in row:
            sentence = f"Tanpri,rakonte yon ti istwa (500 mo) kote wap dekri yon egzanp jounen nan lavi yon pesonaj ki rele {element} epi se yon ayisyen, souple jus rakonte yon istwa a san ou pa rajoute lot kesyon ouyen eksplikasyon"
            writer.writerow([sentence])
```

Prompts for haitian creole Entity Filling

```python
from openpyxl import load_workbook, Workbook
import csv
from collections import defaultdict
input_file = "Prompts_neutral.xlsx"
output_file = "Prompts_neutral_final_creole.csv"

category_map = {
    "Food": "manje",
    "Authors": "otè",
    "Location": "lye",
    "Music": "Mizik",
    "Musical groups": "gwoup mikiz / chantè",
    "Festivals": "fèt",
    "Heroes/Political figures" : "ewo/ figi politik"
}

# Load Excel
wb = load_workbook(input_file)
ws = wb.active

# Write CSV
with open(output_file, mode="w", newline="", encoding="utf-8") as outfile:
    writer = csv.writer(outfile)

    for row in ws.iter_rows(min_row=2, min_col=1, max_col=2, values_only=True):
        category, element = row

        if category and element:
            category_creole = category_map.get(category, category)


        sentence = (
            f"Tanpri, nan fraz sa a ranplase [MASK] pa yon  entite ki soti nan kategori {category_creole} epi ban mwen egzakteman yon sèl antite kòm repons. Pa rekopye fraz la, pa ajoute okenn eksplikasyon oswa detay nan repons lan: "
            f"{element}"
        )

        writer.writerow([sentence])
```

Prompt for french entity filling

```python
from openpyxl import load_workbook, Workbook
import csv
from collections import defaultdict
input_file = "Prompts_neutral.xlsx"
output_file = "Prompts_neutral_final_french.csv"

category_map = {
    "Food": "nourriture",
    "Authors": "autheur",
    "Location": "lieu",
    "Music": "musique",
    "Musical groups": "groupe musique / chanteur",
    "Festivals": "fête",
    "Heroes/Political figures" : "Héros/figures politiques"
}

# Load Excel
wb = load_workbook(input_file)
ws = wb.active

# Write CSV
with open(output_file, mode="w", newline="", encoding="utf-8") as outfile:
    writer = csv.writer(outfile)


    for row in ws.iter_rows(min_row=2, min_col=1, max_col=3, values_only=True):
        category = row[0]   # colonne A
        element = row[2]


        if category and element:
            category_creole = category_map.get(category, category)


        sentence = (
            f"S'il te plait , dans cette phrase , remplace [MASK] par une entité de la catégorie {category_creole} et donne exactement une seule entité comme réponse. Ne recopie pas la phrase, n'ajoute aucune explication ou détails à la réponse : "
            f"{element}"
        )

        writer.writerow([sentence])
```
